```python
def pactorial(n):
  if n==1:
    return 1
  else :
    return n*pactorial(n-1)
n=int(input('1~25의 정수를 입력하세요:'))

for x in range(1,n+1):
  if x<=n:
    print(x,"!=",pactorial(x))
  else:
    break
```
